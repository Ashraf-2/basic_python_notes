```python
# zeors
np4 = np.zeros(10) # point zeros in nth number.   
np4
```

```python
np5 = np.zeros((2,5)) # (dimention, total item in each dimention)
np5
```

```python
np5 = np.zeros((3,5)) # (dimention, total item in each dimention)
np5
```

```python
# full 
np6 = np.full((10),6) # syntex((number of total_items),the item)  
print(np6)
```

```python
# full 
np6 = np.full((5),3) # syntex((number of total_items),the item)  
print(np6)
```

```python
# multidimentional full 
np7 = np.full((2,10),5) # syntex: full((number of dimention, number of items), the item)
np7
```

```python
# multidimentional full 
np7 = np.full((2,5),3) # syntex: full((number of dimention, number of items), the item)
np7
```
