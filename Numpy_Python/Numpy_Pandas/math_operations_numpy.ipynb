Mathematical Operations 

```python
import numpy as np 
```

```python
arr = np.array([30,10,20,50])
```

```python
arr2 = np.array([10,10,10,5]) 
```

```python
print(arr+arr2)
```

```python
np.add(arr,arr2)
```

```python
arr = np.array([30,10,20,50])
arr2 = np.array([10,10,10,5])
print(arr-arr2) 
print(np.subtract(arr2,arr))
print(np.divide(arr,arr2))
```

```python
arr = np.array([30,10,20,50])
arr2 = np.array([2]) 
print(np.pow(arr,arr2))
```
