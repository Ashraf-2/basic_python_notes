## location based indexing `df.loc[row_indexer, column_indexer]`

```python

s1 = pd.Series([12,23,54,21,4,6,78,9])
s1
```

```python
print(s1.loc[[0,2,4]])
```

```python

s2 = pd.DataFrame({
    "name":["python","c","dart", "java","ruby"],
    "pop":[1021,2309,842, 231, 122],
    "rank": [4,1,4,7,3]
} )
s2
```

```python
# get the row 3 and 4
s2.loc[[3,4]]
```

```python
# get the row 1 and 4

s2.loc[[1,4]]
```

```python
# get the row 1 and 4 for the colun 'name' and 'rank'
s2.loc[[1,4], ['name', 'rank']]
```

```python
# get me the all the rows for the column 'name' and 'pop'
s2.loc[:, ['name', 'pop']]
```

```python
# get me the all the rows for the column 'name' and 'pop' where rank is greater than 3
s2.loc[s2['rank'] > 3, ['name', 'pop']]
```

```python
# get the 2nd row and the pop
s2.loc[2,'pop']
```

```python
calories = [420, 380, 390, 450, 500]
s3 = pd.DataFrame(calories)
s3
```

```python
s3.columns = ["calories"]
s3
```

```python
s3.index = ["day1", "day2", "day3", "day4", "day5"]
s3
```

```python
# get the day 4 calories
s3.loc["day4", "calories"]
```

```python
# get the day 4 and 1 calories
s3.loc[["day4", "day1"]]
```

```python
s3.iloc[3] # iloc works in orginal indexing
```
