```python
import matplotlib.pyplot as plt

# Data
names = ['Python', 'C++', 'C', 'Java']
values = [75, 92, 52, 84]
colors = ['b', 'y', 'c', 'g']

# Plot settings
plt.xlabel("Languages")
plt.ylabel("Users")
plt.title("Users - Language Bar Chart")

# Bar chart
plt.bar(names, values, color=colors)

# Display the plot
plt.show()
```

```python
x = [1,2,3,4]
y = [10,20,30,40] 
plt.xlabel('x - axis') 
plt.ylabel('y - axis')
plt.bar(x,y, width=0.4,edgecolor='r',linewidth=5,linestyle=':')
plt.show()
```

```python
x = [1,2,3,4]
y = [10,20,30,40] 
plt.xlabel('x - axis') 
plt.ylabel('y - axis')
plt.bar(x,y, width=0.4,edgecolor='r',linewidth=5,linestyle=':',alpha=0.5, label="popularity")
# plt.legend("no mater")
plt.title("WsCube")

plt.show()
```
