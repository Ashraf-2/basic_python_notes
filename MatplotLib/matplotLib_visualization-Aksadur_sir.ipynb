Creating a figure and a set of subplots -- "Object Oriented" approach

```python
fig, axes = plt.subplots(2,2) # plt.subplots(nrows,ncolumns) 
```

```python
fig, axes = plt.subplots(2,2) # plt.subplots(nrows,ncolumns) 
plt.tight_layout()
```

```python
fig , axes = plt.subplots(2,2, sharex=True, sharey=True, figsize=(12,5)) 
plt.tight_layout()
```

```python
fig , axes = plt.subplots(2,2, sharex=True, sharey=True, figsize=(12,5)) 
# plt.tight_layout()

axes[0][0].plot(x,y,'r--')
axes[0][0].set_xlabel('x-axis')
axes[0][0].set_ylabel('y-axis')
axes[0][0].set_title("Figure-1")

axes[0][1].plot(x,y,'b--')
axes[0][1].set_title("Figure-2")

axes[1][0].plot(x,y,'g--')
axes[1][0].set_title("Figure-3")

axes[1][1].plot(x,y,'c--')
axes[1][1].set_title("Figure-4")
plt.tight_layout()
```
